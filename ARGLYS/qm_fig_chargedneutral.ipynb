```python
import matplotlib.pyplot as plt
import matplotlib as mpl
```

```python
import sys,os
import numpy as np
```

```python
import seaborn as sns
sns.set(style="ticks", color_codes=True, font_scale=1.25)
sns.set_style({"xtick.direction": "in", "ytick.direction": "in"})
```

```python
DGs = np.loadtxt("DGs.dat", comments=['#', '@'])
DGt = np.loadtxt("DGt.dat", comments=['#', '@'])
DDGt = np.loadtxt("DDGt.dat", comments=['#', '@'])
```

```python
import string
import seaborn as sns
sns.set(style="ticks", color_codes=True, font_scale=1.25)
sns.set_style({"xtick.direction": "in", "ytick.direction": "in"})
fig = plt.figure(figsize=(7,6.5))
spec = fig.add_gridspec(2,2)

ax0 = fig.add_subplot(spec[0, :])
ax0.plot(DGs[:,0], DGs[:,1], 'o', mec='k', mfc='C0', linestyle='', label='Lys')
ax0.plot(DGs[:,0], DGs[:,2], 'o', mec='k', mfc='C1', linestyle='', label='Arg')
ax0.plot(DGs[:,0], DGs[:,3], 's', mec='k', mfc='C0', linestyle='', label='Lys$^0$')
ax0.plot(DGs[:,0], DGs[:,4], 's', mec='k', mfc='C1', linestyle='', label='Arg$^0$')
ax0.legend(fontsize=10, ncol=2, columnspacing=0.5, handletextpad=0.1, labelspacing=0.5)
ax0.set_xlim(0,80)
ax0.set_xlabel(r'$\varepsilon$')

ax1 = fig.add_subplot(spec[1,0])
ax1.plot(DGt[:,0], DGt[:,1], 'o', mec='k', mfc='C0', linestyle='', label='Lys')
ax1.plot(DGt[:,0], DGt[:,2], 'o', mec='k', mfc='C1', linestyle='', label='Arg')
ax1.plot(DGt[:,0], DGt[:,3], 's', mec='k', mfc='C0', linestyle='', label='Lys$^0$')
ax1.plot(DGt[:,0], DGt[:,4], 's', mec='k', mfc='C1', linestyle='', label='Arg$^0$')
ax1.legend(fontsize=10, ncol=2, columnspacing=0.5, handletextpad=0.1, labelspacing=0.5)
ax1.set_xlim(0,40)
ax1.set_xlabel(r'$\varepsilon$')

ax2 = fig.add_subplot(spec[1,1])
ax2.plot(DDGt[:,0], DDGt[:,1], 'o', mec='k', mfc='C2', linestyle='', label=r'Lys$\rightarrow$Arg')
ax2.plot(DDGt[:,0], DDGt[:,2], 's', mec='k', mfc='C3', linestyle='', label=r'Lys$^0 \rightarrow$ Arg$^0$')
ax2.set_xlim(0,40)
ax2.legend(fontsize=10, handletextpad=0.1, labelspacing=0.5)

ax0.set_ylabel(r'$\Delta G_\mathrm{Solv}$ (kJ mol$^{-1}$)')
ax1.set_ylabel(r'$\Delta G_\mathrm{Transfer}$ (kJ mol$^{-1}$)')
ax2.set_ylabel(r'$\Delta\Delta G^\mathrm{K\rightarrow R}_\mathrm{Transfer}$ (kJ mol$^{-1}$)')
ax2.set_xlabel(r'$\varepsilon$')
ax1.axhline(linewidth=1, linestyle='--', color='k')
ax2.axhline(linewidth=1, linestyle='--', color='k')

[ax0.text(-0.125, 0.95, string.ascii_uppercase[0], transform=ax0.transAxes, size=20)]
[ax1.text(-0.3, 1, string.ascii_uppercase[1], transform=ax1.transAxes, size=20)]
[ax2.text(-0.3, 1, string.ascii_uppercase[2], transform=ax2.transAxes, size=20)]

plt.tight_layout()

plt.savefig("figures/charged_vs_neutral.png", dpi=300)
```
