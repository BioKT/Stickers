```python
solv_data = pd.read_csv("SOLVENTS.csv")
solv_data
```

```python
solv_data_sel = solv_data.loc[~solv_data['Medium'].isin(['Water', 'Water Phys'])]
display(solv_data_sel)
```

```python
cond_data = pd.read_csv("StickersKR_NEUTRAL.csv")
cond_data
```

```python
combined = pd.concat([ solv_data_sel, cond_data])
display(combined)
```

```python
fig, ax = plt.subplots(figsize=(3.75,4))
colors = ['C0', 'C0', 'C0', 'C1', 'C1', 'C1', 'C1', 'C1', 'Green', 'Purple']
barh = plt.barh(combined['Medium'], combined['ΔΔG BAR'], \
                xerr=combined['ΔΔG Error BAR'],  color=colors, edgecolor='k', linewidth=1.5)
ax.axvline(linewidth=1, linestyle='--', color='k')

ax.set_xlabel('$\Delta \Delta G_\mathrm{Transfer}$ (kJ mol$^{-1}$)')
ax.spines[['right', 'top']].set_visible(False)
ax.set_xlim(-12,2)
ax.set_xticks(range(-10,5,5))
#plt.gca().invert_yaxis()
plt.yticks(fontsize=12)

plt.tight_layout()
plt.savefig("figures/ddg_solvents.png", dpi=300, transparent=True)
```

```python
fig, ax = plt.subplots(figsize=(3.75,4))

ax.errorbar(combined['Epsilon'], combined['ΔΔG BAR'], \
            xerr=combined['Epsilon Err'], yerr=combined['ΔΔG Error BAR'], fmt='', ms=5, ecolor='k', linestyle='')

scatter_kwargs = {"zorder": 100}
ax.scatter(combined['Epsilon'], combined['ΔΔG BAR'], \
            marker='o', c=colors, s=70, edgecolors='k', **scatter_kwargs)

ax.axhline(linewidth=1, linestyle='--', color='k')
ax.set_xlabel(r'$\epsilon$')
ax.set_ylabel('$\Delta \Delta G_\mathrm{Transfer}$ (kJ mol$^{-1}$)')
plt.tight_layout()
plt.savefig("figures/ddg_solvents_vs_epsilon.png", dpi=300, transparent=True)
```
