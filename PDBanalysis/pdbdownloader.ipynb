# Download protein structures from the PDB

```python
import os
import Bio
from Bio.PDB import PDBList
```

```python
culled = "cullpdb_pc15.0_res0.0-1.5_len40-10000_R0.25_Xray_d2025_02_19_chains1916.txt"
with open(culled, 'r') as f:
    raw = f.readlines()
pdbids = [r.split()[0] for r in raw][1:]
print(pdbids)
```

```python
pdbl = PDBList()
for prot in pdbids:
    if not os.path.isfile("PDB/pdb%s.ent"%prot[:4].lower()):
        pdbl.retrieve_pdb_file(prot[:4], file_format='pdb')
```

```python
%%bash 
# Move everything to the folder PDB and remove everything else

for dir in */; do
    if [ "$dir" != "PDB/" ]; then
        echo "${dir}"
        mv "${dir}pdb*ent" PDB/
        rm -r "$dir"
    fi
done
```
